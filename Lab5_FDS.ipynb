```python
# Q23. Convert categorical attributes (Location, Furnishing_Status) into numerical format[cite: 1]
df_encoded = pd.get_dummies(
    df, columns=["Location", "Furnishing_Status"], drop_first=True
)
print("\n--- Q23: Categorical Variables One-Hot Encoded ---")
print(df_encoded.head(2))
```

```python
# SECTION D: OUTLIER ANALYSIS (VISUALIZATION-FOCUSED)
# Q26. Create Box Plots for House Price, Area, Bedrooms[cite: 1]
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
sns.boxplot(y=df["House_Price"], ax=axes[0], color="skyblue").set_title(
    "Q26: House Price"
)
sns.boxplot(y=df["Area"], ax=axes[1], color="lightgreen").set_title("Q26: Area")
sns.boxplot(y=df["Bedrooms"], ax=axes[2], color="salmon").set_title(
    "Q26: Bedrooms"
)
plt.tight_layout()
plt.show()
```

```python

# Q27. Identify outliers visually from boxplots[cite: 1]
print("\n--- Q27: Outliers Visually Identified ---")
print(
    "No data points fall beyond the 1.5x IQR whiskers across House Price, Area, or Bedrooms[cite: 1]."
)

# Q28. Calculate Q1, Q3, IQR for House Price[cite: 1]
Q1 = df["House_Price"].quantile(0.25)
Q3 = df["House_Price"].quantile(0.75)
IQR = Q3 - Q1
print(f"\n--- Q28: Quartiles & IQR for House Price ---")
print(f"Q1: ${Q1:,.2f} | Q3: ${Q3:,.2f} | IQR: ${IQR:,.2f}")

# Q29. Determine Lower Bound and Upper Bound for House Price[cite: 1]
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
print(f"\n--- Q29: Outlier Bounds for House Price ---")
print(
    f"Lower Bound: ${lower_bound:,.2f} | Upper Bound: ${upper_bound:,.2f}[cite: 1]"
)

# Q30. Identify all House Price outliers[cite: 1]
outliers = df[
    (df["House_Price"] < lower_bound) | (df["House_Price"] > upper_bound)
]
print(f"\n--- Q30: Identified Outliers ---")
print(f"Total Outliers Count: {len(outliers)}")

# Q31. Scatter Plot: House Price vs Area to identify unusual observations[cite: 1]
plt.figure(figsize=(7, 4))
sns.scatterplot(data=df, x="Area", y="House_Price", hue="Purchased")
plt.title("Q31: Scatter Plot - House Price vs Area")
plt.xlabel("Area (sq.ft)")
plt.ylabel("House Price ($)")
plt.tight_layout()
plt.show()

```

```python
# Q32. Box Plot for House Price grouped by Location[cite: 1]
plt.figure(figsize=(7, 4))
sns.boxplot(data=df, x="Location", y="House_Price", palette="Set3")
plt.title("Q32: House Price Grouped by Location")
plt.xlabel("Location")
plt.ylabel("House Price ($)")
plt.tight_layout()
plt.show()

```

```python
# Q33. Remove outliers using IQR method[cite: 1]
df_no_outliers = df[
    (df["House_Price"] >= lower_bound) & (df["House_Price"] <= upper_bound)
]
```

```python

# Q34. Compare dataset size before and after outlier removal[cite: 1]
print("\n--- Q34: Size Comparison ---")
print(
    f"Size Before: {len(df)} rows | Size After: {len(df_no_outliers)} rows[cite: 1]"
)

# Q35. Discuss impact of outliers on house price prediction models[cite: 1]
print("\n--- Q35: Impact of Outliers ---")
print(
    "Outliers artificially tilt linear regression best-fit lines, inflate Mean Squared Error metrics,[cite: 1]"
)
print("and skew decision tree split boundaries[cite: 1].")
```
