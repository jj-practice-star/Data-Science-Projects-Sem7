```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import StandardScaler
from scipy import stats
```

```python
df = pd.read_csv("student-mat.csv", sep=';')
display(df.head(10))
print('------------------Last 10 recrods-------------------')
display(df.tail(10))
print('------------------Shape-------------------')
print("Shape:", df.shape)
print('------------------Columns-------------------')
print(df.columns)
print('------------------Datatypes-------------------')
print(df.dtypes)
print('------------------summary-------------------')
print(df.describe())
```

```python
print(df.isnull().sum())
print('------------------duplicates check-------------------')
print("Duplicate Rows:", df.duplicated().sum())
print('------------------Identify Categorical and Numerical Columns-------------------')
categorical_cols = df.select_dtypes(include='object').columns
numerical_cols = df.select_dtypes(exclude='object').columns
print(categorical_cols)
print(numerical_cols)
```

```python
le = LabelEncoder()
df_encoded = df.copy()
for col in categorical_cols:
    df_encoded[col] = le.fit_transform(df_encoded[col])
```

```python
df_onehot = pd.get_dummies(df,
columns=categorical_cols,drop_first=True)
```

```python
scaler = MinMaxScaler()
df_norm = df_encoded.copy()
df_norm[numerical_cols] = scaler.fit_transform(
df_norm[numerical_cols]
)
```

```python
std_scaler = StandardScaler()
df_std = df_encoded.copy()
df_std[numerical_cols] = std_scaler.fit_transform(
df_std[numerical_cols]
)
```

```python
df['AgeGroup'] = pd.cut(
df['age'],
bins=[0,16,18,100],
labels=['Teenager',
'Young Adult',
'Adult']
)
```

```python
def performance(g):
  if g >= 16:
    return "Excellent"
  elif g >= 12:
    return "Good"
  elif g >= 8:
    return "Average"
  else:
    return "Poor"
df['Performance'] = df['G3'].apply(performance)
print(df[['G3','Performance']].head())
```

```python
Q1 = df['absences'].quantile(0.25)
Q3 = df['absences'].quantile(0.75)

IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers = df[
(df['absences'] < lower) |
(df['absences'] > upper)
]
print("Total Outliers:",
len(outliers))
```

```python
Q1 = df['G3'].quantile(0.25)
Q3 = df['G3'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers = df[
(df['G3'] < lower) |
(df['G3'] > upper)
]
print("Outliers in G3:",
len(outliers))
```

```python
df_clean = df.copy()
for col in ['age','absences','G1','G2','G3']:
  Q1 = df_clean[col].quantile(0.25)
  Q3 = df_clean[col].quantile(0.75)
  IQR = Q3 - Q1
  lower = Q1 - 1.5 * IQR
  upper = Q3 + 1.5 * IQR
  df_clean = df_clean[
  (df_clean[col] >= lower) &
  (df_clean[col] <= upper)
  ]
```

```python
print("Before:", df.shape)
print("After :", df_clean.shape)
```

```python
removed = (
(len(df)-len(df_clean))
/ len(df)
) * 100
print(round(removed,2),"%")
```

```python
z = np.abs(stats.zscore(
df[['age','absences','G1','G2','G3']]))
outliers_z = df[
(z > 3).any(axis=1)
]
print("Z-score Outliers:",
len(outliers_z))
```
